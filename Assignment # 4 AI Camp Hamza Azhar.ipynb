```python
students = {
    "Ali": [78, 85, 90, 88, 76],
    "Burhan": [65, 70, 72, 68, 75],
    "Hamza": [90, 92, 88, 95, 91]
}
for name, marks in students.items():
    total = sum(marks)
    percentage = (total / (len(marks) * 100)) * 100
    print(name, "Percentage:", percentage)
```
